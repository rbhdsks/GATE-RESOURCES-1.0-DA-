## GATE 2018

```python
def fun(n):
    i = n 
    j = 0
    sum = 0
    while i > 1:
        j += 1
        i //= 2
    while j > 1:
        j //= 2
        sum += 1

    return sum
```
What is the return value of the function fun when it is called with the argument 2⁴⁰?  

ans = 5

## GATE 2003

```python
def find_func(x, y):
    p = 1
    s = 1
    for i in range(1, y):
        p *= x/i
        s += p
    return s
```
For large values of y, the return value of the function find_func best approximates:  
a. x^y  b. e^x  c. ln(1+x)  d. x^x

ans = e^x

## GATE 

```python
def exp(X, Y):
    res = 1
    a = X
    b = Y

    while b != 0:
        if b % 2 == 0:
            a = a * a
            b = b // 2
        else:
            res = res * a
            b = b - 1

    return res
```
The function exp(X, Y) computes X^Y for positive integers X and Y. Which of the following conditions is true before every iteration of the while loop?  
a. X^Y = a^b  
b. (res * a)^Y = (res * X)^b  
c. X^Y = res * a^b  
d. X^Y = (res * a)^b  

opt. c

The value returned by myX is  
a. the maximum possible sum of elements in any sub-list of E  
b. the maximum element in any sub-list of E  
c. sum of the maximum elements in all sub-lists of E  
d. sum of all elements in E  


opt. a
