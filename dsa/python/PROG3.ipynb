Multi-dimensional lists:
<br>
These are lists within lists. 

````python
l = [[1, 2, 3, [4, 5]], [23, [12, 34], 9], 30, 21]
print(f"len(l): {len(l)}")
print(f"l[0]: {l[0]}")
# to access 4
print(f"{l[0][3][0]} or {l[0][3][-2]} or {l[-4][3][0]}")
````

````python
l = [1, 2, 3, 4]
print(f"{l}, \tid(l): {id(l)}")
l.append(5)
print(f"{l}, \tid(l): {id(l)}")
# id doesn't change after append
l.append([6, 7])
print(l)
l.extend([8, 9])
print(l)
l.insert(0, 0)  # list.insert(index, value)
print(l)
l.insert(2, 1.5)
print(l)
l.insert(1000, 10)  # if index is out of range, it will insert at the end
print(l)
l.insert(-1, 9.5)  # insert at the second last index
print(l)
l.insert(-1000, -1)  # if index is out of range, it will insert at the beginning
print(l)
````

Summary:
- append: list.append(x) adds x to the end of the list, returns None <br>
- extend: list.extend(iterable) adds the elements of the iterable to the end of the list, returns None <br>
- insert: list.insert(i, x) inserts x at index i, returns None <br>
- remove: list.remove(x) removes the first occurrence of x, returns None <br>
- pop: list.pop(i) removes the element at index i, returns the element <br>
- del: del list[i] removes the element at index i, returns None <br>
- clear: list.clear() removes all elements from the list, returns None <br>
- index: list.index(x) returns the index of the first occurrence of x <br>
- count: list.count(x) returns the number of occurrences of x <br>
- sort: list.sort() sorts the list in place, returns None    <br>
- reverse: list.reverse() reverses the list in place, returns None <br>
- copy: list.copy() returns a shallow copy of the list <br>
shallow copy: a new list is created and the elements are added to it. If the elements are mutable, the references are copied, not the elements themselves. <br>
- list comprehension: [x for x in iterable if condition] <br>
- list comprehension with nested loops: [x for x in iterable1 for y in iterable2 if condition] <br>

list comprehension 

````python
l = [i for i in range(10)]
print(l)
````

````python
l = [i**2 for i in range(10)]
print(l)
````

````python
a = [1, 2, 3]
b = ['a', 'b', 'c']
cross_product = [(i, j) for i in a for j in b]
print(cross_product)
````

````python
# using if-else in list comprehension
l = [i if i % 2 == 0 else i**2 for i in range(10)]
# square the number if it is odd, else keep it as it is
print(l)
````

````python
a = [1, 2, 3]
b = [4, 5, 6]

l = [(i, j) if i+j>5 else (j, i) for i in a for j in b]
print(l)
````

shallow vs deep copy

````python
sub = ['a', 'b']

l = [1, 2, sub]
m = l.copy()

sub[0:] = 'hello', 'there'

print(m)
````

````python
import copy

sub = ['a', 'b']

l = [1, 2, sub]
m = copy.deepcopy(l)

sub[0:] = 'hello', 'there'

print(m)
````

### sets

````python
s = set()
# s = {} is an empty dict, not a set
s = {1, 2, 3} # s = {1} is a set as well
````

Sets are unordered, so they have no indexing. Sets are unhashable. 

Sets don't allow other mutable objects inside them. Lists, dictionaries, or other sets can't be present inside sets. 

````python
s = {1, 2, 3, [4, 5]}
````

Tuples can be elements of sets.

````python
t = (4, 5, 6)
s = {1, 2, 3, t}
s
````

A tuple containing a list can't be an element of a set.

````python
{1, 2, 3, (4, 5, [6, 7])}
````

Sets don't allow duplicates

````python
{1, 2, 3, 4, 3, 4, 5}
````

Sets can't be indexed, but we can check whether an element is present in a set.

````python
'obi-wan' in {1, 2, 3, 'obi-wan', 4, 5}
````

````python
s = {1, 2}
s.add(3)
s
````

set.update(iterable) unions the set with the iterable and puts it all into the set.

````python
s = {1, 2, 3}
s.update([1, 2, 3, 4, 5, 6])
s
````

set.union(iterable) doesn't update the set itself, it just returns the union.

````python
print(s.union([7, 8, 9]))
print(s)
````

````python
s = set()
s.update("hello there")
s
````

set.discard(key) would remove the key from the set if present.
<br>
set.remove(key) would remove the key if present, otherwise throw an error.  

````python
s = {1, 2, 3, 4, 5}
s.remove(3)
s
````

````python
s.remove(10)
````

````python
s.discard(10)
s
````

set.pop() takes no arguments and removes a random element from the set.

````python
s = set("hello there")
print(s)
s.pop()
print(s)
````

#### set operations

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
````

````python
# union
print(a | b)
print(a.union(b))
````

````python
# intersection 
print(a & b)
print(a.intersection(b))
````

A difference between these functions and the operators is that the operators only take sets as arguments.
``` set1 | set2 ``` is fine, but ``` list1 | set1 ``` is not.
Both ``` set1.union(set2) ``` and ``` set1.union(list1) ``` are fine.
``` list1.union(set1) ``` is not fine, since the List object as no such member function.

````python
# union update
set1 = {1, 2, 4}
set2 = {4, 5}

set1 |= set2
print(set1)

# similarly there's set1 &= set2
# set1.intersection_update(set2)
````

````python
# set difference
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}

print(f"{set1} - {set2} = {set1 - set2}")

# difference update
set1 -= set2
# set1.difference_update(set2)
print(set1)
````

Symmetric difference is essentially the union minus the intersection.

````python
# symmetric difference
set1 = {'abc', 'pqr', 'xyz', 'lmn'}
set2 = {'alpha', 'beta', 'abc', 'pqr', 'gamma', 'delta'}

print(set1.symmetric_difference(set2))
print((set1 | set2) - (set1 & set2))
print(set1 ^ set2)
````

````python
# symmetric difference update
set1 = {'abc', 'pqr', 'xyz', 'lmn'}
set2 = {'alpha', 'beta', 'abc', 'pqr', 'gamma', 'delta'}

set1.symmetric_difference_update(set2)
# set1 ^= set2
print(set1)
````

````python
set1 = {1, 2, 3}
set2 = set1.copy()
set2
````

The isdisjoint function tells if the 2 elements have no elements in common. The issuperset and issubset functions work correspondingly.

````python
set1 = {1, 2, 3, 4, 5}
set2 = {5, 6, 7, 8, 9}
set3 = {10, 20, 30}

print(set1.isdisjoint(set2))
print(set1.isdisjoint(set3))
````

````python
set1 = {1, 2, 3, 4, 5}
set2 = {3, 4, 5}
set3 = {5, 6}

print(set1.issuperset(set2))
print(set1.issuperset(set3))
print(set2.issubset(set1))
print(set3.issubset(set1))
````

#### Frozen sets
These are just immutable sets. 
They don't have member functions like ``` add() ``` or ``` remove() ```.
They follow all the properties followed by sets though. 

````python
f = frozenset([1, 2, 3, True, False])
f
````

Set operations like union, intersection, and set difference can be done on frozen sets. Union update and the other updates won't work. Copy works.
