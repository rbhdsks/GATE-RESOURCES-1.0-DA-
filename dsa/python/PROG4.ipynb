# Dictionaries

## Dictionaries

#### Initialization

````python
d = {
    'a': '1',
    'b': '2',
    'c': '3',
    'd': '4',
}
print(d)
````

````python
d = dict(a='1', b='2', c=[3, 3.5, 3.7], d='4')
print(d)
````

````python
d = dict([('a', '1'), ('b', '2'), ('c', '3'), ('d', '4')])
print(d)
````

#### Properties

Dictionaries are ordered.

````python
d = dict(zip(['a', 'b', 'c', 'd'], ['1', '2', '3', '4']))
print(d)
d = dict(zip(['b', 'c', 'a', 'd'], ['2', '3', '1', '4']))
print(d)
````

But they cannot be indexed by integers.
Instead, they are indexed by keys.

````python
print(d['a'])
````

The keys must be hashable (meaning they must be immutable). Non-unique keys are overwritten.

````python
# this is invalid
d = {
    's': 1
    [1, 2]: 2,  # SyntaxError: invalid syntax
    3: 3
    's': 4      # will overwrite the previous 's'
}
````

````python
d = {
    1: 'a', 
    2: 'b',
    0: 'c',
    1.0: 'c',
    1.1: 'd',
    False: 'd',
}
print(d)  
````

A dictionary can be copied using the `copy` method.

````python
d1 = { 'a': 1, 'b': 2 }
d2 = d1.copy()
# here changing d1 will not affect d2
d1['a'] = 3
print(d2)
print(id(d1), id(d2))
````

A dictionary can be created using the fromkeys method.

````python
keyList = ['a', 'b', 'c', 'd']

d = dict.fromkeys(keyList)
print(d)
````

````python
keyList = ['a', 'b', 'c', 'd']

d = dict.fromkeys(keyList, 0)
print(d)
````

#### Keys, values, items

````python
d = {
    'hello': 'there',
    'general': 'kenobi',
    'you': 'are',
    'a': 'bold',
    'one': '!'
}

print(d.get('hello'))
print(d.get('missing_key'))
print(d.get('missing_key', 'default_value'))
````

````python
print(d.items())
````

````python
print(d.keys())
````

````python
print(d.values())
````

````python
for k, v in d.items():
    print(k, v)
````

#### Modifying a dictionary

````python
d = {
    'hello': 'there',
    'general': 'kenobi',
    'you': 'are',
    'a': 'bold',
    'one': '!'
}
````

````python
d['hello'] = 'hi'
d
````

The ```dict.pop(key)``` method removes the key and returns the value. It doesn't work without an argument. <br>The ```dict.popitem()``` method removes the last key-value pair and returns it. <br>The ```dict.clear()``` method removes all key-value pairs.

````python
d.pop('hello')
````

````python
d
````

````python
d.pop("random_bullshit_go")
````

To avoid a KeyError in ```dict.pop(key)```, you can pass a default value as a second argument as ```dict.pop(key, default_value)```. <br>To avoid a KeyError in ```dict[key]```, you can use the ```dict.get(key)``` method.

````python
d.pop('missing_key', 'default_value')
````

````python
d.setdefault('hello', 'there')
````

````python
d
````

````python
d = {
    'hello': 'there',
    'general': 'kenobi',
    'you': 'are',
    'a': 'bold',
    'one': '!'
}
````

````python
d.popitem()
````

````python
d
````

``` dict.update() ``` essentially merges two dictionaries into the first. It returns None. It acts kind of like the `extend` method for lists, or like union update in set theory.

````python
d = {
    'hello': 'there',
    'general': 'kenobi',
    'you': 'are',
    'a': 'bold',
    'one': '!'
}
````

````python
d.update({'new_key': 'new_value'})
d
````

It also accepts a list of tuples. Like ``` dict.update([('a', 1), ('b', 2)]) ```.<br>
Or even a list of lists. Like ``` dict.update([['a', 1], ['b', 2]]) ```.<br>
Or even a string of tuples. Like ``` dict.update(a=1, b=2) ```.

````python
d.update(i='dont',know='what', im='doing')
````
