## GATE 2016


```python
def f(p: list, n: int) -> int:
    if n <= 1:
        return 0
    else:
        return max(f(p[1:], n-1), p[0]-p[1])

a = [3, 5, 2, 6, 4]
print(f(a, 5))
```
What is the output of the above code snippet?

ans = 3  


## GATE 2015


Suppose c = (c[0], ..., c[k-1]) is a list of length k, where all the entries are from the set {0, 1}. For any +ve integers a and n, consider the following pseudocode.

```
DOSOMETHING(c, a, n):
    z = 1
    for i in range(0, k):
        do z = z^2 mod n
        if c[i] = 1:
            do z = z*a mod n
    return z
```

If k=4, c = (1, 0, 1, 1), a=2 and n=8, the output of DOSOMETHING(c, a, n) is ______.

ans = 0  



## GATE 2015

```python
x = 10
def f1():
    x = 25
    x += 1
    return x

def f2():
    x = 50
    def inner_f2():
        nonlocal x
        x += 1
        return x
    return inner_f2()

def f3(y=x):
    z = y*10
    return z

x = 1
call_f2 = f2()
x += f1() + call_f2() + f3() + call_f2()
print(x)
```
What is the output of the above code snippet?

ans = 230  


## GATE 2015



```python
def get(n):
    if n < 1:
        return
    else:
        get(n-1)
        get(n-3)
```
If `get(6)` is called, how many times will `get()` be invoked?

ans = 25

Consider the following pseudocode. What is the total number of multiplications to be performed?

```
D = 2
for i = 1 to n do
    for j = i to n do
        for k = j+1 to n do
            D = D * 3
```

a. half of the product of the 3 consecutive integers  
b. one-third of the product of the 3 consecutive integers  
c. one-sixth of the product of the 3 consecutive integers  
d. none of the above

ans = option c
